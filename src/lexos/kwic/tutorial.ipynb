# Key Words In Context (KWIC)

Lexos includes a module called `kwic`, which allows users to search for a keyword and then returns all instances of that keyword and the surrounding context of each keyword. This module is a wrapper class for the implementation of kwic found in the textacy library.

This tutorial shows how to use the `kwic` class to find keywords in context.


## Setup

```python
from lexos.kwic import Kwic
from spacy.tokens import Doc
```

## Searching for Keywords

You can use the `kwic` class to search a string for the context of keywords in either a string or a spacy doc.

### Searching a spaCy Doc

The `kwic` allows for users to search a spaCy doc, which is an object type from the library `spaCy`. If a user has a string or some other text file that they wish to convert to a spaCy doc before searching, they can use the `tokenizer` module to generate a doc. 

The method `find()` is used to search for keywords. The method returns a list of all instances of said keyword, the preceding context, and the context following the keyword.

```python
from lexos.tokenizer import Tokenizer

text = "If you are looking to find keywords in context of a spaCy doc, then look no further. The kwic module will soon become your best friend!"
tokenizer = Tokenizer()
doc = tokenizer(text)
kwic_results = Kwic.find(doc=doc, keyword="spaCy", window_size=50, pad_context=False)
results_list = list(kwic_results)
print(results_list)
```

You can also use a regex expression to search for multiple variations of a keyword or phrase.

```python
text = "This is a test string with multiple variations of the word Test, which will be searched using a regex expression."
doc = tokenizer(text)
kwic_results = Kwic.find(doc=doc, keyword=r"[Tt]est", window_size=50, pad_context=False)
results_list = list(kwic_results)
print(results_list)
```

Finally, you can also search just a text string instead of a spaCy doc.

```python
text = "If you are looking to find keywords in context of a spaCy doc, then look no further. The kwic module will soon become your best friend!"
kwic_results = Kwic.find(doc=text, keyword="spaCy", window_size=50, pad_context=False)
results_list = list(kwic_results)
print(results_list)
```

### Search Parameters

There are multiple parameters that can be changed in the `find()` method to customize the output:


- **ignore_case**: Weather to ignore the case of the keywords in the search. Like demonstrated with the regex keyword, if this parameter is set to 'True', then a search of the keyword 'test' will return hits of 'test', 'Test', 'TEST', etc.
- **window_size**: The size of the context to be returned in characters. For example, a window size of 50 will return the preceding and following 50 characters from the position of the keyword.
- **pad_context**: When set to 'True', the returned contexts will always be the size of the window, regardless of their position in the text. If there is not enough characters to fill out the window size, then leading/following spaces will be added depending on if it is at the beginning or end of the text, respectivley.

## Search Variations

The `kwic` class includes multiple variations of the `find` method that users can use to get the exact output they need.

### `find_to_dataframe`

The `find_to_dataframe()` method functions the exact same way as the `find()` method, but returns the results of the search as a Pandas dataframe instead of a list.

It shares all the same parameters.

```python
text = "If you are looking to find keywords in context of a spaCy doc, then look no further. The kwic module will soon become your best friend!"
kwic_results = Kwic.find_to_dataframe(doc=text, keyword="spaCy", window_size=50, pad_context=False)
kwic_results
```

### `find_multiple_keywords()`

This method allows for a user to search for multiple keywords in a kwic search.

All parameters are the same except for `keywords`, which now takes a list of strings, as opposed to a singular string. Each object in the return list now also contains the keyword that that result relates to as the last instance in the list.

```python
text = "If you are looking to find keywords in context of a spaCy doc, then look no further. The kwic module will soon become your best friend!"
kwic_results = Kwic.find_multiple_keywords(doc=text, keywords=["spaCy", "kwic"], window_size=50, pad_context=False)
results_list = list(kwic_results)
print(results_list)
```

### `find_in_sentences()`

This methods returns only the sentences that the keyword is in as context.

This method only takes in spaCy docs as input, no strings. It still contains the `ignore_case` parameter, but does not contain the `pad_context` or `window_size` parameters, since the context is predetermined by the sentence(s) that the keyword presides in.

```python
text = "What's the deal with airline food? It's like they don't even try! I mean, the food is terrible, the seats are cramped, and the service is non-existent. But hey, at least they give you a tiny bag of pretzels to munch on while you suffer through the flight!"
doc = tokenizer(text)
kwic_results = Kwic.find_in_sentences(doc=doc, keyword="food", ignore_case=True)
results_list = list(kwic_results)
print(results_list)
```

### `find_tokens()`

This method is a combination of the `find()` and `find_in_sentences()` methods. This method allows for a user to search for keywords in context and select the size of the context in tokens. 

Like `find_in_sentences()`, this method only accepts spaCy docs as input for the `doc` parameter. This method uses the parameter `token_window` to determine how big the context window is, in tokens. 

```python
text = "What's the deal with airline food? It's like they don't even try! I mean, the food is terrible, the seats are cramped, and the service is non-existent. But hey, at least they give you a tiny bag of pretzels to munch on while you suffer through the flight!"
doc = tokenizer(text)
kwic_results = Kwic.find_tokens(doc=doc, keyword="food", token_window=5, ignore_case=True)
results_list = list(kwic_results)
print(results_list)
```
